```python
# Imports and file path
from pathlib import Path
import pandas as pd
import time

PROJECT_ROOT = Path.cwd().parent

RAW_FILE = (
    PROJECT_ROOT
    / "Raw_Data"
    / "NFHS_5_India_Districts_Factsheet_Data.xls"
)

if not RAW_FILE.exists():
    raise FileNotFoundError(f"Raw file not found: {RAW_FILE}")

print(f"Source file: {RAW_FILE.name}")
```

```python
# Inspect workbook sheets
workbook = pd.ExcelFile(RAW_FILE, engine="xlrd")
display(pd.DataFrame({"Worksheet": workbook.sheet_names}))
```

```python
# Load the source data
start_time = time.perf_counter()

df_raw = pd.read_excel(
    RAW_FILE,
    sheet_name="Sheet1",
    engine="xlrd"
)

print(f"Rows: {len(df_raw):,}")
print(f"Columns: {len(df_raw.columns):,}")
print(f"Load time: {time.perf_counter() - start_time:.2f} seconds")

display(df_raw.head())
```

```python
# Inspect columns and data types
column_profile = pd.DataFrame({
    "column_name": df_raw.columns,
    "data_type": df_raw.dtypes.astype(str).values,
    "missing_values": df_raw.isna().sum().values,
    "unique_values": df_raw.nunique().values,
})

display(column_profile)
```

```python
# Baseline quality checks
print(f"Missing values: {df_raw.isna().sum().sum():,}")
print(f"Duplicate rows: {df_raw.duplicated().sum():,}")

for column in ["District Names", "State/UT"]:
    if column in df_raw.columns:
        print(f"Missing {column}: {df_raw[column].isna().sum():,}")

display(df_raw[["District Names", "State/UT"]].head(10))
```
